```python
import matplotlib.pyplot as plt
import cv2
import pandas as pd
import os

# Load metadata (expects you already ran generator/main.py)
df = pd.read_csv('../output/metadata.csv')
df = df.head(10).reset_index(drop=True)

coco_images_dir = "/home/harshita/coco/train2017"
tampered_dir = "../output/tampered"
mask_dir = "../output/binary_masks"

# Show 10 samples: rows=10, cols=3 (Original / Tampered / Mask)
fig, axs = plt.subplots(len(df), 3, figsize=(15, 4 * len(df)))

for i, sample in df.iterrows():
    tampered_path = os.path.join(tampered_dir, sample["image_filename"])
    orig_path = os.path.join(coco_images_dir, sample["image_filename"].replace(".png", ".jpg"))
    mask_path = os.path.join(mask_dir, sample["mask_filename"])

    cmfd_bgr = cv2.imread(tampered_path)
    orig_bgr = cv2.imread(orig_path)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

    cmfd = cv2.cvtColor(cmfd_bgr, cv2.COLOR_BGR2RGB) if cmfd_bgr is not None else None
    orig = cv2.cvtColor(orig_bgr, cv2.COLOR_BGR2RGB) if orig_bgr is not None else None

    axs[i, 0].imshow(orig if orig is not None else [[0]])
    axs[i, 0].set_title(f"{i+1}. Original")
    axs[i, 0].axis("off")

    axs[i, 1].imshow(cmfd if cmfd is not None else [[0]])
    axs[i, 1].set_title("Tampered")
    axs[i, 1].axis("off")

    axs[i, 2].imshow(mask if mask is not None else [[0]], cmap="gray")
    axs[i, 2].set_title("Tri-Map Mask")
    axs[i, 2].axis("off")

plt.tight_layout()
plt.show()
```
